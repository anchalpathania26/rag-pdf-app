```python
## creating a simple txt file
import os
os.makedirs("../data/text_files", exist_ok = True)
```

```python
sample_text={

"../data/text_files/python_basics.txt": """python programming introduction

Python is a high level programming language that is easy to read and write. It was created by Guido van Rossum and is widely used by beginners as well as professionals. Python focuses on simplicity, which makes it one of the best languages to start learning programming.

Python is popular because its syntax is clean and similar to everyday English. It allows developers to write fewer lines of code compared to many other languages. It also has a large community and a vast number of libraries that help in different areas like web development, data analysis, automation, and artificial intelligence.

In Python, you can store data using variables. There are different types of data such as text, numbers, and true or false values. You can control the flow of your program using conditions, which help you make decisions, and loops, which allow you to repeat tasks multiple times.

Functions are an important concept in Python. They allow you to write reusable blocks of code so you do not have to repeat the same logic again and again. This makes programs more organized and easier to maintain.

Python also supports libraries, which are pre written code modules that help you perform complex tasks easily. For example, some libraries help with data handling, some with creating graphs, and others with building AI applications.

Python is used in many fields such as web development, data science, machine learning, automation, and software development. Because of its simplicity and power, it is one of the most in demand programming languages today.

In simple terms, Python is a versatile and beginner friendly language that helps you build everything from small scripts to advanced intelligent systems.

"""
}
```

```python
for filepath, content in sample_text.items():
    with open(filepath, 'w', encoding = 'utf-8') as f:
        f.write(content)

        print("Sample text files created!")
```

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/text_files/python_basics.txt", encoding="utf-8")
document = loader.load()
print(document)
```

```python
### Directory Loader ( we can try loading the content with the help of directory also)
from langchain_community.document_loaders import DirectoryLoader
## loading all the text files from the directory
dir_loader = DirectoryLoader(
    path="../data/text_files",
    glob="**/*.txt", ## Pattern to match the file
    loader_cls=TextLoader, # loader class to use 
    loader_kwargs= {"encoding": "utf-8"}, # additional arguments for the loader class
    show_progress=False # do not show progress while loading the files
)

documents = dir_loader.load()
print(documents)
```

```python
### Now we are loading the pdfs
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader
## loading all the pdf files from the directory
dir_loader = DirectoryLoader(
    path="../data/Pdfs",
    glob="**/*.Pdf", ## Pattern to match the file
    loader_cls=PyMuPDFLoader, # loader class to use 
    show_progress=False # do not show progress while loading the files
)

pdf_documents = dir_loader.load()
print(pdf_documents)
```
